Pipeline: tokenize → lowercase → remove punctuation → remove stopwords → sorted unique base list →
Porter / Snowball / WordNet lemmatizer / 4-grams.

```python
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
from nltk.util import ngrams

STOP_WORDS = set(stopwords.words("english"))
PUNCTUATION = set(string.punctuation)
```

## Part A – Preprocessing (base cleaned list)

```python
corpus = (
    "Language is a method of communication with the help of which we can speak, read and write. "
    "Natural Language Processing (NLP) is the sub field of computer science especially Artificial "
    "Intelligence (AI) that is concerned about enabling computers to understand and process human "
    "language. We have various open-source NLP tools but NLTK (Natural Language Toolkit) scores very "
    "high when it comes to the ease of use and explanation of the concept. The learning curve of "
    "Python is very fast and NLTK is written in Python so NLTK is also having very good learning kit. "
    "NLTK has incorporated most of the tasks like tokenization, stemming, Lemmatization, Punctuation, "
    "Character Count, and Word count. It is very elegant and easy to work with."
)


def preprocess(text):
    """Tokenize, lowercase, drop punctuation and stopwords; return sorted unique list."""
    tokens = word_tokenize(text)                                   # 1. tokenize
    tokens = [tok.lower() for tok in tokens]                       # 2. lowercase
    tokens = [tok for tok in tokens if tok not in PUNCTUATION]     # 3. remove punctuation
    tokens = [tok for tok in tokens if tok not in STOP_WORDS]      # 4. remove stopwords
    return sorted(set(tokens))                                     # 5. unique + sorted


base_cleaned_list = preprocess(corpus)
print("Base Cleaned List:")
print(base_cleaned_list)
print("Length:", len(base_cleaned_list))
```

## Set 1 – Porter Stemmer

```python
porter = PorterStemmer()
set1_porter = sorted(set(porter.stem(word) for word in base_cleaned_list))
print("Set 1 (Porter Stemmer):")
print(set1_porter)
```

## Set 2 – Snowball Stemmer (English)

```python
snowball = SnowballStemmer("english")
set2_snowball = sorted(set(snowball.stem(word) for word in base_cleaned_list))
print("Set 2 (Snowball Stemmer):")
print(set2_snowball)
```

## Set 3 – WordNet Lemmatizer

```python
lemmatizer = WordNetLemmatizer()
set3_lemma = sorted(set(lemmatizer.lemmatize(word) for word in base_cleaned_list))
print("Set 3 (WordNet Lemmatizer):")
print(set3_lemma)
```

## Set 4 – 4-grams

```python
four_grams = list(ngrams(base_cleaned_list, 4))
print("Set 4 (4-grams):")
for gram in four_grams:
    print(" ".join(gram))
n = len(base_cleaned_list)
print(f"\nTotal 4-grams generated: {len(four_grams)}  (n - 3 = {n} - 3 = {n - 3})")
```

## Additional Task 2 – Porter vs Snowball

```python
words_2 = ['running', 'happily', 'generously', 'argued', 'believing',
           'organization', 'studies', 'friendship', 'beautifully', 'connected']

print(f"{'Word':<15}{'Porter Stem':<15}{'Snowball Stem':<15}")
print(f"{'-'*6:<15}{'-'*11:<15}{'-'*13:<15}")
different = []
for word in words_2:
    p_stem, s_stem = porter.stem(word), snowball.stem(word)
    print(f"{word:<15}{p_stem:<15}{s_stem:<15}")
    if p_stem != s_stem:
        different.append((word, p_stem, s_stem))

if different:
    print("\nWords with different stems:")
    for word, p_stem, s_stem in different:
        print(f"  {word}: Porter={p_stem}, Snowball={s_stem}")
else:
    print("\nWords with different stems: None for this list.")
```
